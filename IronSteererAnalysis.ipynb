```python
#Dependencies 
import uproot
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Unpacking the ROOT file and getting the magnetic fields
FilePath = "Output/IronSteerer_sim_Output.root"
StepData = uproot.open(FilePath+":component_step_world_DATA")
Positions = StepData.arrays(["position_x","position_y","position_z","time"],library="np")
x = Positions["position_x"]
y = Positions["position_y"]
z = Positions["position_z"]
MagneticFields = StepData.arrays(["magnetic_field_x","magnetic_field_y","magnetic_field_z"],library="np")
magx = MagneticFields["magnetic_field_x"]
magy = MagneticFields["magnetic_field_y"]
magz = MagneticFields["magnetic_field_z"]
```

```python
#Determining the magnetic field at the center of the coils
for i in range(len(magz)):
    if x[i] == y[i] == 0:
        print(magz[i]*10000)
        break
```
